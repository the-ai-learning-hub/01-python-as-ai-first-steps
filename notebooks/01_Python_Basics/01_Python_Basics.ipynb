# 2. Variáveis e tipos

<a id="sec-2-1"></a>

## 2.1. Atribuição

Uma **variável** é um nome que guarda um valor. Criamo-la com `=` (lê-se "recebe"):

```python
nome_da_variavel = valor
```

````python
idade = 25
nome = "Ana"
print(nome, "tem", idade, "anos")
````

Uma variável pode mudar de valor: a atribuição nova substitui a anterior. Também podemos usar o valor antigo para calcular o novo.

````python
pontos = 10
pontos = pontos + 5   # o lado direito é calculado primeiro: 10 + 5
print(pontos)
````

````python
# Atribuição múltipla: vários nomes de uma vez
x, y = 3, 7
print(x, y)

# Trocar os valores de duas variáveis
x, y = y, x
print(x, y)
````

> ⚠️ **Cuidado:** em Python, `=` **não** é "igual" no sentido matemático: é uma ordem para guardar um valor. Para *perguntar* se dois valores são iguais usa-se `==` (secção 6).

<a id="sec-2-2"></a>

## 2.2. Regras para nomes de variáveis

- Só letras, algarismos e `_` (sem espaços nem hífenes).
- **Não podem começar por um número**: `nota1` ✅, `1nota` ❌.
- **Distinguem maiúsculas de minúsculas** (*case-sensitive*): `idade`, `Idade` e `IDADE` são três variáveis diferentes.
- Não podem ser **palavras reservadas** do Python, como `if`, `for`, `def`, `True` ou `class`.
- Por convenção, em Python escrevemos os nomes em **snake_case**: minúsculas e palavras separadas por `_` (`preco_total`, `nome_cliente`).
- Escolhe nomes que digam o que a variável guarda: `preco_total` é muito melhor do que `x2`.

````python
idade = 30
Idade = 99
print(idade, Idade)   # são variáveis diferentes
````

````python
1nota = 15   # SyntaxError: um nome não pode começar por um número
````

````python
# Lista das palavras reservadas (não as podes usar como nomes)
import keyword
print(keyword.kwlist)
````

<a id="sec-2-3"></a>

## 2.3. Tipos básicos

Cada valor tem um **tipo**, que determina o que se pode fazer com ele:

| Tipo | Nome | Exemplos |
|---|---|---|
| `int` | número inteiro | `1`, `-7`, `2026` |
| `float` | número decimal (vírgula flutuante) | `3.14`, `-0.5`, `2.0` |
| `str` | texto (*string*) | `"Olá"`, `'Python'` |
| `bool` | valor lógico | `True`, `False` |
| `NoneType` | "nada", ausência de valor | `None` |

Para descobrir o tipo de um valor usamos `type()`:

````python
print(type(42))
print(type(3.14))
print(type("Olá"))
print(type(True))
print(type(None))
````

> 📝 **Nota:** em Python as casas decimais escrevem-se com **ponto**, não com vírgula: `3.14`, nunca `3,14`.

Quando misturamos `int` e `float` numa conta, o resultado é `float`:

````python
a = 2      # int
b = 2.5    # float
resultado = a + b * 4
print(resultado, type(resultado))
````

O valor `None` representa "não há valor". Vais encontrá-lo bastante, por exemplo quando uma função não devolve nada (secção 4.3).

````python
telefone = None   # ainda não sabemos o número
print(telefone)
````

<a id="sec-2-4"></a>

## 2.4. Conversões entre tipos

As funções `int()`, `float()`, `str()` e `bool()` convertem valores de um tipo para outro:

````python
print(int("42") + 1)     # texto → inteiro: 43
print(float("3.5") * 2)  # texto → decimal: 7.0
print(str(42) + "!")     # inteiro → texto: 42!
print(float(8))          # inteiro → decimal: 8.0
````

Ao converter um decimal para inteiro com `int()`, o Python **trunca**: corta a parte decimal, **não arredonda**.

````python
print(int(7.3))   # 7
print(int(7.9))   # 7  (não é 8!)
print(int(-7.9))  # -7
````

Com `bool()`, o zero e o texto vazio dão `False`; quase tudo o resto dá `True` (mais sobre isto na secção 6.4):

````python
print(bool(0), bool(5), bool(""), bool("olá"))
````

Nem todas as conversões são possíveis:

````python
int("quarenta e dois")   # ValueError: este texto não representa um número
````

> ⚠️ **Cuidado:** a função `input()`, que usaste na Lição 00, **devolve sempre uma string**, mesmo que o utilizador escreva um número. Para fazer contas tens de converter:

> print(type(idade))                      # <class 'str'>

> ```

# 3. Operadores

<a id="sec-3-1"></a>

## 3.1. Operadores aritméticos

| Operador | Operação | Exemplo | Resultado |
|---|---|---|---|
| `+` | adição | `7 + 2` | `9` |
| `-` | subtração | `7 - 2` | `5` |
| `*` | multiplicação | `7 * 2` | `14` |
| `/` | divisão (dá sempre `float`) | `7 / 2` | `3.5` |
| `//` | divisão inteira | `7 // 2` | `3` |
| `%` | resto da divisão inteira | `7 % 2` | `1` |
| `**` | potência | `7 ** 2` | `49` |

````python
a = 10
b = 3

print(a + b)   # 13
print(a - b)   # 7
print(a * b)   # 30
print(a / b)   # 3.3333333333333335
print(a // b)  # 3
print(a % b)   # 1
print(a ** b)  # 1000
````

> 💡 **Dica:** o resto `%` é muito útil: `n % 2 == 0` diz-te se `n` é par, e `n % 10` dá-te o último algarismo de `n`.

> 📝 **Nota:** repara que `10 / 3` dá `3.3333333333333335` e não `3.3333333333333333`. Os computadores guardam decimais em binário, com uma pequena imprecisão. Por isso `0.1 + 0.2` dá `0.30000000000000004`. É normal e acontece em quase todas as linguagens.

````python
print(0.1 + 0.2)
print(6 / 2)      # mesmo uma divisão exata dá float: 3.0
````

<a id="sec-3-2"></a>

## 3.2. Precedência e parênteses

Como na matemática, as operações têm uma ordem: primeiro `**`, depois `*`, `/`, `//`, `%`, e por fim `+` e `-`. Os **parênteses** mudam essa ordem e tornam o código mais legível.

````python
print(2 + 3 * 4)      # 14: a multiplicação vem primeiro
print((2 + 3) * 4)    # 20: os parênteses vêm primeiro
print(2 * 3 ** 2)     # 18: a potência vem antes da multiplicação
print(-2 ** 2)        # -4: a potência vem antes do sinal negativo!
print((-2) ** 2)      # 4
````

> 💡 **Dica:** na dúvida, põe parênteses. Ninguém se queixa de código claro.

## 3.3. Atribuição aumentada

É muito comum atualizar uma variável a partir do seu próprio valor. Há atalhos para isso:

| Atalho | Equivale a |
|---|---|
| `x += 5` | `x = x + 5` |
| `x -= 5` | `x = x - 5` |
| `x *= 5` | `x = x * 5` |
| `x /= 5` | `x = x / 5` |

````python
saldo = 100
saldo += 50    # depósito
saldo -= 30    # levantamento
saldo *= 2     # duplicou!
print(saldo)   # 240
````

<a id="sec-3-4"></a>

## 3.4. Funções numéricas úteis

O Python traz funções prontas a usar:

- `round(x)` — arredonda ao inteiro; `round(x, n)` — arredonda a `n` casas decimais;
- `abs(x)` — valor absoluto (tira o sinal);
- `min(...)` e `max(...)` — o menor e o maior de vários valores.

````python
print(round(3.14159, 2))   # 3.14
print(round(7.8))          # 8
print(abs(-12))            # 12
print(min(4, 9, 1, 7))     # 1
print(max(4, 9, 1, 7))     # 9
````

> ⚠️ **Cuidado:** `round()` usa o **arredondamento bancário** (*round half to even*): quando o número está exatamente a meio (`.5`), arredonda para o **inteiro par** mais próximo. Assim, `round(2.5)` dá `2` e não `3`! Isto evita que, somando muitos arredondamentos, o erro acumule sempre para cima.

````python
print(round(0.5))   # 0
print(round(1.5))   # 2
print(round(2.5))   # 2  ← surpresa!
print(round(3.5))   # 4
````

# 4. Funções

<a id="sec-4-1"></a>

## 4.1. Definir e chamar uma função

Uma **função** é um bloco de código com nome, que faz uma tarefa e pode ser reutilizado. Já usaste várias: `print()`, `type()`, `round()`… Agora vais criar as tuas.

```python
def nome_da_funcao(parametros):
    # corpo da função (indentado com 4 espaços)
    ...
```

- `def` diz ao Python que vamos **definir** uma função;
- o corpo tem de estar **indentado** (4 espaços);
- definir uma função não a executa: para isso temos de a **chamar**, escrevendo o nome seguido de `()`.

````python
def saudacao():
    print("Olá, bem-vindo ao Python!")

saudacao()   # chamar a função
saudacao()   # e outra vez
````

<a id="sec-4-2"></a>

## 4.2. Parâmetros e `return`

Os **parâmetros** são as "entradas" da função. A instrução `return` define a "saída": o valor que a função **devolve** a quem a chamou.

````python
def saudar(nome):
    print("Olá,", nome)

saudar("Ana")
saudar("Rui")
````

````python
def dobro(x):
    return 2 * x

resultado = dobro(3)    # o valor devolvido fica guardado
print(resultado)        # 6
print(dobro(10) + 1)    # 21: podemos usar o resultado noutras contas
````

Uma função pode ter vários parâmetros, separados por vírgulas:

````python
def area_retangulo(largura, altura):
    return largura * altura

print(area_retangulo(3, 4))   # 12
````

> 📝 **Nota:** quando o Python encontra um `return`, a função **termina** imediatamente. O código que estiver depois não é executado.

<a id="sec-4-3"></a>

## 4.3. `return` vs `print`

É a confusão mais comum de quem está a começar:

- `print()` **mostra** um valor no ecrã, mas esse valor perde-se;
- `return` **devolve** o valor, para o podermos guardar e usar.

Uma função **sem `return` devolve `None`**:

````python
def soma_com_print(a, b):
    print(a + b)

def soma_com_return(a, b):
    return a + b

x = soma_com_print(2, 3)    # mostra 5...
y = soma_com_return(2, 3)   # não mostra nada...
print("x =", x)             # ...mas x ficou None!
print("y =", y)             # ...e y ficou 5
````

> ⚠️ **Cuidado:** se vires `None` a aparecer onde não esperavas, verifica se a tua função tem `return`. E se tentares fazer contas com esse `None` (`x + 1`), vais ter um `TypeError`.

<a id="sec-4-4"></a>

## 4.4. Valores por omissão e argumentos com nome

Um parâmetro pode ter um **valor por omissão**, usado quando não indicamos nada. Ao chamar a função podemos também indicar os argumentos **pelo nome**, em qualquer ordem.

````python
def preco_com_iva(preco, iva=0.23):
    return preco * (1 + iva)

print(preco_com_iva(100))             # usa o IVA por omissão (23%)
print(preco_com_iva(100, 0.06))       # IVA reduzido (6%)
print(preco_com_iva(iva=0.13, preco=100))  # argumentos com nome
````

> 📝 **Nota:** repara em `123.0` e `106.0`: como `0.23` é `float`, o resultado também é. E `112.99999999999999` (em vez de `113.0`): lá está a imprecisão dos decimais (secção 3.1). Quando quisermos mostrar valores destes, formatamo-los (secção 5.6).

## 4.7. Âmbito das variáveis (*scope*)

As variáveis criadas **dentro** de uma função são **locais**: só existem enquanto a função está a ser executada e não se veem cá fora. As variáveis criadas fora de qualquer função são **globais** e podem ser lidas dentro das funções.

````python
taxa = 0.23          # global

def calcular_iva(valor):
    iva = valor * taxa   # 'iva' é local; 'taxa' é lida da variável global
    return iva

print(calcular_iva(100))
print(taxa)
````

````python
print(iva)   # NameError: 'iva' só existia dentro da função
````

> 💡 **Dica:** se uma função precisa de um valor, o mais claro é recebê-lo como **parâmetro** e devolver o resultado com `return`, em vez de depender de variáveis globais.

## 5.3. Indexação e *slicing*

Cada caracter tem uma posição (**índice**), a **começar em 0**. Também podemos contar a partir do fim com índices **negativos**:

```
 P   y   t   h   o   n
 0   1   2   3   4   5
-6  -5  -4  -3  -2  -1
```

````python
palavra = "Python"
print(palavra[0])    # P  (primeiro)
print(palavra[3])    # h
print(palavra[-1])   # n  (último)
print(palavra[-2])   # o  (penúltimo)
````

O ***slicing*** extrai um pedaço: `texto[inicio:fim]` vai do índice `inicio` **até `fim`, sem o incluir**. Se omitires o início, começa no 0; se omitires o fim, vai até ao fim. Um terceiro valor, `texto[inicio:fim:passo]`, salta caracteres.

````python
palavra = "Python"
print(palavra[0:3])   # Pyt  (índices 0, 1, 2)
print(palavra[:2])    # Py   (do início até ao 2, exclusive)
print(palavra[2:])    # thon (do 2 até ao fim)
print(palavra[-3:])   # hon  (os 3 últimos)
print(palavra[::2])   # Pto  (de 2 em 2)
print(palavra[::-1])  # nohtyP (passo -1: string invertida!)
````

> 💡 **Dica:** `texto[:n]` dá os primeiros `n` caracteres e `texto[-n:]` os últimos `n`. Um *slice* fora dos limites não dá erro (`"abc"[:100]` é `"abc"`), mas um índice simples dá: `"abc"[10]` → `IndexError`.

Em vez disso, criamos uma **nova** string (e, se quisermos, guardamo-la na mesma variável):

O operador `in` verifica se um texto aparece dentro de outro (e `not in` o contrário). Distingue maiúsculas de minúsculas.

````python
frase = "Diogo Freitas"
print("Freitas" in frase)    # True
print("freitas" in frase)    # False (f minúsculo)
print("z" not in frase)      # True
````

## 5.6. Formatação: `.format()` e f-strings

Para montar texto com valores lá dentro, há duas formas. A mais antiga é o método `.format()`, que substitui cada `{}` por um valor:

````python
print("Eu adoro {}, isto é mesmo {}!".format("Python", "fixe"))
````

A forma moderna (e recomendada) são as **f-strings**: pões um `f` antes das aspas e escreves as variáveis ou expressões diretamente entre `{}`.

````python
animal = "papagaio"
nome_animal = "Shakira"
a = 1
b = 2
print(f"Tenho um {animal} chamado {nome_animal}. Sabias que {a} + {b} é igual a {a + b}?")
````

Depois de `:` dentro das chavetas podes indicar o **formato**:

| Formato | Significado | Exemplo | Resultado |
|---|---|---|---|
| `{x:.2f}` | 2 casas decimais | `f"{3.14159:.2f}"` | `3.14` |
| `{n:,}` | separador de milhares | `f"{1234567:,}"` | `1,234,567` |
| `{p:.1%}` | percentagem | `f"{0.256:.1%}"` | `25.6%` |
| `{s:>8}` | alinhar à direita em 8 posições | `f"{'ok':>8}"` | `      ok` |

````python
preco = 112.99999999999999   # o resultado de preco_com_iva(100, 0.13)
populacao = 10639726
print(f"Preço: {preco:.2f} €")
print(f"População: {populacao:,} habitantes")
print(f"Taxa de sucesso: {0.256:.1%}")
````

> 🤖 **Ligação à IA:** as f-strings são a forma mais prática de montar *prompts* para modelos de linguagem: `f"Resume este texto em {n} frases: {texto}"`.

# 6. Booleans, comparações e lógica

Um **booleano** (`bool`) só tem dois valores possíveis: `True` (verdadeiro) e `False` (falso). São o resultado de perguntas como "é maior?" ou "é igual?", e são a base de todas as decisões de um programa.

<a id="sec-6-1"></a>

## 6.1. Operadores de comparação

| Operador | Pergunta |
|---|---|
| `==` | é igual a? |
| `!=` | é diferente de? |
| `<` / `>` | é menor / maior que? |
| `<=` / `>=` | é menor / maior ou igual a? |

````python
x = 5
y = 10

print(x < y)    # True  (5 é menor que 10)
print(x > y)    # False
print(x == y)   # False
print(x != y)   # True
print(x >= 5)   # True
print(x <= 4)   # False
````

````python
# Também funciona com texto (ordem alfabética; maiúsculas vêm antes das minúsculas)
print("ana" == "ana")    # True
print("Ana" == "ana")    # False
print("abacate" < "banana")  # True
````

> ⚠️ **Cuidado:** `=` guarda um valor; `==` compara dois valores. Escrever `if x = 5:` dá `SyntaxError`.

<a id="sec-6-2"></a>

## 6.2. Comparações encadeadas

Em Python podes escrever intervalos tal como em matemática:

````python
nota = 14
print(0 <= nota <= 20)    # True: equivale a (0 <= nota) and (nota <= 20)
print(10 < nota < 12)     # False
````

<a id="sec-6-3"></a>

## 6.3. `and`, `or` e `not`

- `a and b` — `True` só se **ambos** forem verdadeiros;
- `a or b` — `True` se **pelo menos um** for verdadeiro;
- `not a` — inverte: `True` passa a `False` e vice-versa.

| `a` | `b` | `a and b` | `a or b` | `not a` |
|---|---|---|---|---|
| `True` | `True` | `True` | `True` | `False` |
| `True` | `False` | `False` | `True` | `False` |
| `False` | `True` | `False` | `True` | `True` |
| `False` | `False` | `False` | `False` | `True` |

````python
idade = 20
tem_bilhete = True

print(idade >= 18 and tem_bilhete)   # True: pode entrar
print(idade < 12 or idade >= 65)     # False: não tem desconto
print(not tem_bilhete)               # False
````

## 6.4. Verdadeiro e falso (*truthiness*)

Qualquer valor pode ser usado como condição. São considerados **falsos**: `0`, `0.0`, a string vazia `""`, `None` (e `False`, claro). Quase tudo o resto é **verdadeiro**.

````python
print(bool(0), bool(0.0), bool(""), bool(None))   # tudo False
print(bool(42), bool(-1), bool("0"), bool(" "))   # tudo True (até "0" e " "!)
````

É por isso que podes escrever, por exemplo, `if nome:` para verificar se uma string não está vazia (vês isto na secção 7).

# 7. Decisões: `if` / `elif` / `else`

Até agora, todos os nossos programas executavam todas as linhas, de cima para baixo. Com `if` o programa passa a **tomar decisões**.

<a id="sec-7-1"></a>

## 7.1. `if` e `else`

```python
if condicao:
    # executado se a condição for verdadeira
else:
    # executado se for falsa
```

- A linha do `if` termina com **dois pontos** `:`;
- o bloco de código a executar está **indentado** (4 espaços). É a indentação que diz ao Python onde o bloco começa e acaba;
- o `else` é opcional.

````python
temperatura = 28

if temperatura > 25:
    print("Está calor!")
    print("Leva água.")       # também pertence ao if (está indentado)
else:
    print("Está fresquinho.")

print("Esta linha é sempre executada")   # já não está indentada
````

````python
nome = ""

if nome:    # truthiness: uma string vazia é falsa
    print(f"Olá, {nome}!")
else:
    print("Não indicaste o teu nome.")
````

> ⚠️ **Cuidado:** a indentação não é decorativa. Se te esqueceres dela, o Python dá `IndentationError`. Usa sempre 4 espaços (o editor fá-lo por ti quando carregas em `Tab`).

<a id="sec-7-2"></a>

## 7.2. `elif`: várias alternativas

`elif` (abreviatura de *else if*) testa uma nova condição **só se as anteriores forem falsas**. Podes ter quantos `elif` quiseres; assim que um for verdadeiro, os seguintes são ignorados. O `else` final apanha "tudo o resto".

````python
idade = 20

if idade < 18:
    print("Menor de idade")
elif idade < 65:    # aqui já sabemos que idade >= 18
    print("Adulto")
else:
    print("Sénior")
````

````python
palavra = "Python"

if len(palavra) < 5:
    print("Palavra curta")
elif len(palavra) <= 8:
    print("Palavra de comprimento médio")
else:
    print("Palavra longa")
````

````python
letra = "A"

if letra.isupper():
    print("Letra maiúscula")
elif letra.islower():
    print("Letra minúscula")
else:
    print("Não é uma letra (algarismo, símbolo ou espaço)")
````

<a id="sec-7-3"></a>

## 7.3. `if` dentro de `if`

Podemos pôr um `if` dentro de outro (*nested if*). Cada nível acrescenta mais 4 espaços de indentação.

````python
idade = 25
tem_carta = False

if idade >= 18:
    if tem_carta:
        print("Pode conduzir.")
    else:
        print("Tem idade, mas precisa de tirar a carta.")
else:
    print("Ainda não tem idade para conduzir.")
````

> 💡 **Dica:** muitas vezes um `if` aninhado pode ser simplificado com `and`. Mas não exageres: se ficar mais difícil de ler, mantém os níveis.

<a id="sec-7-4"></a>

## 7.4. A ordem das condições importa

Como o Python pára na **primeira** condição verdadeira, a ordem dos testes é decisiva. Este código tem um erro: um 19 é classificado como "Positiva"!

````python
nota = 19

# ❌ Ordem errada: a primeira condição apanha tudo o que é >= 10
if nota >= 10:
    print("Positiva")
elif nota >= 18:
    print("Excelente")   # nunca chega aqui!
````

````python
nota = 19

# ✅ Ordem certa: do caso mais exigente para o menos exigente
if nota >= 18:
    print("Excelente")
elif nota >= 10:
    print("Positiva")
else:
    print("Negativa")
````

<a id="sec-7-5"></a>

## 7.5. `match` / `case`

Quando comparamos **a mesma variável** com uma lista de valores possíveis, o Python (3.10 ou mais recente) oferece `match` / `case`, que fica mais arrumado do que uma longa cadeia de `elif`. O `case _` apanha "tudo o resto" (como o `else`), e `|` significa "ou".

````python
def nome_do_dia(numero):
    match numero:
        case 1:
            return "Segunda-feira"
        case 2:
            return "Terça-feira"
        case 3 | 4 | 5:
            return "A meio da semana"
        case 6 | 7:
            return "Fim de semana 🎉"
        case _:
            return "Dia inválido"

print(nome_do_dia(1))   # Segunda-feira
print(nome_do_dia(4))   # A meio da semana
print(nome_do_dia(7))   # Fim de semana 🎉
print(nome_do_dia(9))   # Dia inválido
````

> 📝 **Nota:** `match` compara valores concretos. Para intervalos (`nota >= 18`, `idade < 65`…), o `if` / `elif` continua a ser a escolha natural.

## 7.6. Curiosidade: decimal para binário

Os computadores guardam tudo em **binário** (só zeros e uns). O Python tem funções prontas para converter: `bin()` dá o binário de um número (com o prefixo `0b`) e `int(texto, 2)` faz o caminho inverso.

````python
def decimal_para_binario(n):
    """Devolve a representação binária de n, sem o prefixo '0b'."""
    return bin(n)[2:]   # bin(5) dá '0b101'; o slicing tira o '0b'

print(f"Decimal 3 em binário: {decimal_para_binario(3)}")
print(f"Decimal 4 em binário: {decimal_para_binario(4)}")
print(f"Binário 101 em decimal: {int('101', 2)}")
````

```python
print(validar_password("Passw0rd!"))    # True
```

Dar um nome a cada regra (variáveis booleanas) torna o código fácil de ler e de alterar. No notebook 03 vais aprender ciclos, que permitem verificar cada caracter um a um (por exemplo, exigir mesmo um algarismo).
</details>
