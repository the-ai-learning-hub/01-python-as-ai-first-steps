Até agora escreveste todo o código no próprio notebook. Mas ninguém programa assim no mundo real: **reutilizamos código** escrito por outras pessoas (ou por nós próprios, noutro ficheiro). É para isso que servem os módulos.

<a id="sec-1-1"></a>

## 1.1. O que é um módulo, um pacote e uma biblioteca?

| Termo | O que é | Exemplo |
|---|---|---|
| **Módulo** | Um ficheiro `.py` com funções, variáveis e classes | `math`, `random`, `utilidades.py` |
| **Pacote** | Uma pasta com vários módulos organizados | `json`, `collections`, `pandas` |
| **Biblioteca** | Nome informal para um conjunto de módulos/pacotes que resolvem um problema | pandas, matplotlib, requests |
| **Biblioteca padrão** | Os módulos que já vêm com o Python — não precisas de instalar nada | `math`, `datetime`, `json` |

Para usares um módulo, tens de o **importar** primeiro.

<a id="sec-1-2"></a>

## 1.2. Formas de importar

**1) Importar o módulo inteiro** — usas `modulo.nome` para aceder ao que está lá dentro:

````python
import math

print(math.sqrt(16))   # raiz quadrada → 4.0
print(math.pi)         # o número π
````

**2) Importar só o que precisas** com `from … import …` — usas o nome diretamente:

````python
from math import sqrt, pi

print(sqrt(25))   # 5.0
print(round(pi, 2))   # 3.14
````

**3) Dar uma alcunha (alias)** com `as` — muito usado para nomes compridos. Há convenções que toda a gente segue:

````python
import pandas as pd              # convenção: pandas → pd
import matplotlib.pyplot as plt  # convenção: matplotlib.pyplot → plt

print(pd.__version__)   # a versão do pandas instalada
````

> ⚠️ **Cuidado:** evita `from math import *`. Importa *tudo* para o teu código sem dizer o quê — fica difícil saber de onde vem cada nome, e podes sobrescrever variáveis tuas sem dares por isso.

> 💡 **Dica:** coloca os `import` no início do ficheiro/notebook. Assim, quem lê vê logo de que é que o código depende.

## 1.3. Uma visita à biblioteca padrão

O Python vem com "pilhas incluídas": centenas de módulos prontos a usar. Estes são os mais úteis para começar.

### 1.3.1. `math` — matemática

````python
import math

print(math.sqrt(81))       # 9.0
print(math.ceil(2.1))      # arredonda para cima → 3
print(math.floor(2.9))     # arredonda para baixo → 2
print(math.factorial(5))   # 5! = 120
print(round(math.pi, 4))   # 3.1416
````

### 1.3.2. `random` — números aleatórios

````python
import random

print(random.randint(1, 6))                    # um inteiro entre 1 e 6 (inclusive)
print(random.random())                         # um decimal entre 0 e 1
print(random.choice(["pedra", "papel", "tesoura"]))  # um elemento ao acaso

baralho = [1, 2, 3, 4, 5]
random.shuffle(baralho)   # baralha a lista (altera-a!)
print(baralho)
````

Executa a célula acima várias vezes: os resultados mudam. Às vezes queremos o contrário — **resultados reprodutíveis** (iguais sempre que se executa). Para isso usamos `random.seed()`:

````python
random.seed(42)
primeira = [random.randint(1, 100) for _ in range(5)]

random.seed(42)   # mesma semente...
segunda = [random.randint(1, 100) for _ in range(5)]

print(primeira)
print(segunda)
print(primeira == segunda)   # ...mesmos números → True
````

> 🤖 **Ligação à IA:** em machine learning fixa-se sempre uma semente (`seed` ou `random_state`) para que as experiências possam ser repetidas e comparadas.

### 1.3.3. `datetime` — datas e horas

````python
from datetime import date, datetime

hoje = date.today()
print("Hoje é", hoje)   # muda conforme o dia em que executares

inicio = date(2026, 1, 1)
natal = date(2026, 12, 25)
print((natal - inicio).days)            # dias entre as duas datas → 358
print(natal.strftime("%d/%m/%Y"))       # formatar → 25/12/2026

texto = "2026-03-15 14:30"
momento = datetime.strptime(texto, "%Y-%m-%d %H:%M")   # texto → data
print(momento.year, momento.month, momento.hour)       # 2026 3 14
````

### 1.3.4. `statistics` — estatística básica

````python
import statistics

notas = [7, 8, 8, 9, 10, 6]
print(statistics.mean(notas))              # média → 8
print(statistics.median(notas))            # mediana → 8.0
print(statistics.mode(notas))              # moda (valor mais frequente) → 8
print(round(statistics.stdev(notas), 2))   # desvio padrão → 1.41
````

### 1.3.5. `collections.Counter` — contar coisas

Lembras-te de contar elementos com um dicionário e um ciclo `for`? O `Counter` faz isso numa linha.

````python
from collections import Counter

frutas = ["maçã", "pera", "maçã", "banana", "pera", "maçã"]
contagem = Counter(frutas)

print(contagem)                  # Counter({'maçã': 3, 'pera': 2, 'banana': 1})
print(contagem["maçã"])          # 3
print(contagem["kiwi"])          # 0 (não dá erro!)
print(contagem.most_common(1))   # [('maçã', 3)]
print(Counter("banana"))         # também conta letras → Counter({'a': 3, 'n': 2, 'b': 1})
````

### 1.3.6. `json` — trocar dados com outros programas

**JSON** é o formato de texto mais usado para trocar dados na web. Parece-se muito com um dicionário Python. O módulo `json` converte nos dois sentidos:

- `json.loads(texto)` → texto JSON para **dicionário/lista** Python
- `json.dumps(objeto)` → dicionário/lista Python para **texto** JSON

````python
import json

# Imagina que isto é a resposta que recebeste de uma API (é só texto!)
resposta = '{"pergunta": "Qual é a capital de Portugal?", "resposta": "Lisboa", "confianca": 0.97, "fontes": ["wikipedia", "atlas"]}'
print(type(resposta))   # <class 'str'>

dados = json.loads(resposta)   # texto → dicionário
print(type(dados))             # <class 'dict'>
print(dados["resposta"])       # Lisboa
print(dados["fontes"][0])      # wikipedia
````

````python
aluno = {"nome": "João", "idade": 21, "cursos": ["Python", "Dados"]}

texto_json = json.dumps(aluno, ensure_ascii=False, indent=2)   # dicionário → texto bonito
print(texto_json)
````

> 🤖 **Ligação à IA:** as APIs de modelos de linguagem (LLMs) respondem em JSON. O primeiro passo é quase sempre `json.loads(...)` para teres um dicionário com que trabalhar.

## 2.3. `if __name__ == "__main__":`

Todo o módulo tem uma variável especial, `__name__`:

- Quando o **importas**, `__name__` é o nome do módulo (`"utilidades"`).
- Quando o **executas diretamente** (`uv run python utilidades.py`), `__name__` é `"__main__"`.

````python
print(utilidades.__name__)   # utilidades
print(__name__)              # o notebook é o programa "principal" → __main__
````

Por isso, o código dentro de `if __name__ == "__main__":` **só corre quando executas o ficheiro diretamente**, e não quando o importas. É o sítio ideal para testes rápidos. Repara que, quando fizemos `import utilidades`, nada foi impresso.

# 3. Erros e exceções

Errar faz parte de programar — até os profissionais veem erros todos os dias. Quando algo corre mal, o Python **lança uma exceção**: pára o programa e mostra uma mensagem a explicar o que aconteceu. Aprender a ler essas mensagens poupa-te horas.

> 📝 **Nota:** as células desta secção que dão erro **são de propósito**. Lê a mensagem e segue em frente.

<a id="sec-3-1"></a>

## 3.1. Ler um traceback

A mensagem de erro chama-se **traceback**. Mostra o "caminho" que o programa percorreu até rebentar. A regra de ouro:

> 💡 **Dica:** **lê de baixo para cima.** A última linha diz o **tipo** de erro e a **mensagem**. As linhas acima mostram **onde** aconteceu (a seta `---->` aponta para a linha culpada).

````python
def calcular_media(notas):
    return sum(notas) / len(notas)

def relatorio(aluno, notas):
    media = calcular_media(notas)
    return f"{aluno}: média {media:.1f}"

print(relatorio("Rui", []))   # o Rui ainda não tem notas...
````

A última linha diz `ZeroDivisionError: division by zero`. Subindo, vemos que aconteceu em `calcular_media`, chamada por `relatorio`, chamada na última linha da célula. A lista vazia fez `len(notas)` ser `0`.

<a id="sec-3-2"></a>

## 3.2. Os erros mais comuns

| Erro | Quando acontece | Exemplo |
|---|---|---|
| `SyntaxError` | O código não é Python válido (falta `:`, `)`, `"`…) | `print("olá"` |

| `TypeError` | Operação com o tipo errado | `"idade: " + 25` |

| `ZeroDivisionError` | Divisão por zero | `10 / 0` |

| `KeyError` | Chave que não existe num dicionário | `{"a": 1}["b"]` |

| `FileNotFoundError` | O ficheiro não existe | `open("nao_existe.txt")` |

Vamos ver cada um a acontecer. O `SyntaxError` é especial: o Python nem chega a executar a célula, porque não a consegue perceber.

## 3.3. `try` / `except`

Muitas vezes **sabemos** que algo pode falhar (o utilizador escreve texto em vez de um número, um ficheiro não existe, a internet cai…). Em vez de deixar o programa rebentar, **apanhamos** a exceção:

```python
try:
    # código que pode falhar
except TipoDeErro:
    # o que fazer se falhar
```

````python
texto = "42a"

try:
    numero = int(texto)
    print("Convertido:", numero)   # esta linha não chega a correr
except ValueError:
    print(f"'{texto}' não é um número válido.")

print("O programa continua normalmente ✅")
````

Com `as` guardas a exceção numa variável e podes mostrar a mensagem original:

````python
try:
    numero = int("42a")
except ValueError as erro:
    print("Ups! Erro:", erro)   # Ups! Erro: invalid literal for int() with base 10: '42a'
````

<a id="sec-3-4"></a>

## 3.4. Apanhar exceções específicas

Podes ter vários `except`, um para cada tipo de erro. O Python usa o **primeiro** que corresponder.

````python
def dividir(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Não é possível dividir por zero."
    except TypeError:
        return "Os dois valores têm de ser números."

print(dividir(10, 4))     # 2.5
print(dividir(10, 0))     # Não é possível dividir por zero.
print(dividir(10, "2"))   # Os dois valores têm de ser números.
````

> ⚠️ **Cuidado:** nunca escrevas um `except:` sozinho (sem tipo). Apanha **todos** os erros — incluindo os teus erros de escrita e até o "parar" do Jupyter — e esconde os problemas em vez de os resolver. Apanha sempre o erro específico que esperas. Se mesmo assim precisares de um "apanha-tudo", usa `except Exception as erro:` e mostra o `erro`.

<a id="sec-3-5"></a>

## 3.5. `else` e `finally`

- `else` → corre **só se não houve erro** no `try`.
- `finally` → corre **sempre**, com ou sem erro (ótimo para "arrumar a casa": fechar ficheiros, ligações, …).

> 📝 **Nota:** a função `open()` abre um ficheiro — vais aprender a trabalhar com ficheiros no notebook 06.

````python
def primeira_linha(caminho):
    try:
        ficheiro = open(caminho, encoding="utf-8")
    except FileNotFoundError:
        print(f"❌ O ficheiro '{caminho}' não existe.")
    else:
        print("✅ Primeira linha:", ficheiro.readline().strip())
        ficheiro.close()
    finally:
        print(f"   (terminei de tentar ler '{caminho}')")

primeira_linha("utilidades.py")
primeira_linha("nao_existe.txt")
````

## 3.6. Lançar os teus erros: `raise`

Também podes ser tu a lançar uma exceção, com `raise`. É a forma correta de uma função dizer "**recuso-me a continuar com estes dados**" — muito melhor do que devolver um valor esquisito em silêncio.

`isinstance(valor, tipo)` verifica se um valor é de um certo tipo (`True`/`False`).

````python
def definir_idade(idade):
    if not isinstance(idade, int):
        raise TypeError("A idade tem de ser um número inteiro.")
    if idade < 0 or idade > 150:
        raise ValueError(f"Idade inválida: {idade}")
    return idade

print(definir_idade(30))   # 30

for valor in [-5, "trinta", 200]:
    try:
        definir_idade(valor)
    except (TypeError, ValueError) as erro:   # um tuplo apanha vários tipos de uma vez
        print(f"{type(erro).__name__}: {erro}")
````

## 3.7. Exemplo prático: converter texto em números

Dados reais vêm "sujos": valores em falta, texto onde devia haver números, vírgulas em vez de pontos… Vamos converter uma lista de strings em números **sem que um valor mau estrague tudo**.

````python
valores = ["12", "7.5", "abc", "", "3", "1,5", " 42 "]

numeros = []
invalidos = []

for valor in valores:
    try:
        numeros.append(float(valor.replace(",", ".")))   # aceita vírgula decimal
    except ValueError:
        invalidos.append(valor)

print("Números:", numeros)       # [12.0, 7.5, 3.0, 1.5, 42.0]
print("Inválidos:", invalidos)   # ['abc', '']
print("Soma:", sum(numeros))     # 66.0
````

> 🤖 **Ligação à IA:** limpar dados é grande parte do trabalho em IA. O pandas tem funções que fazem isto em massa (por exemplo, `pd.to_numeric(..., errors="coerce")`), mas a ideia é exatamente esta.

## 3.8. Tentar outra vez: o padrão *retry*

Quando chamas um serviço na internet (uma API), às vezes ele falha por razões temporárias: servidor ocupado, rede lenta… A solução habitual é **tentar outra vez**, algumas vezes, com uma pausa entre tentativas.

Vamos simular uma API "instável" que falha nas duas primeiras chamadas:

````python
import time

estado = {"chamadas": 0}

def chamar_api():
    """Finge ser uma API instável: falha nas 2 primeiras chamadas."""
    estado["chamadas"] += 1
    if estado["chamadas"] < 3:
        raise ConnectionError("servidor ocupado")
    return {"estado": "ok", "texto": "Olá! Sou um modelo de linguagem."}


def chamar_com_retry(funcao, max_tentativas=5, espera=0.2):
    for tentativa in range(1, max_tentativas + 1):
        try:
            return funcao()                  # correu bem → sai logo da função
        except ConnectionError as erro:
            print(f"Tentativa {tentativa} falhou ({erro}). A tentar outra vez...")
            time.sleep(espera)               # espera um pouco antes de tentar de novo
    raise RuntimeError(f"Desisti depois de {max_tentativas} tentativas.")


resposta = chamar_com_retry(chamar_api)
print(resposta["texto"])
````

> 🤖 **Ligação à IA:** é exatamente assim que se chamam APIs de LLMs em produção. Normalmente a espera **duplica** a cada tentativa (`espera *= 2`), para não sobrecarregar o servidor — chama-se *exponential backoff*.

# 4. Classes e objetos

Uma **classe** é um molde que junta **dados** (atributos) e **comportamento** (métodos) num só sítio. Um **objeto** (ou instância) é uma coisa concreta criada a partir desse molde.

Pensa numa classe `Aluno`: todos os alunos têm nome e notas (dados) e todos sabem calcular a sua média (comportamento). A Ana e o Rui são dois **objetos** diferentes da mesma classe.

<a id="sec-4-1"></a>

## 4.1. Já usas objetos!

Em Python, **tudo é um objeto**: textos, números, listas, dicionários… Quando escreves `valor.algo()`, estás a chamar um **método** desse objeto.

````python
texto = "olá mundo"
lista = [3, 1, 2]

print(type(texto))      # <class 'str'>  → texto é um objeto da classe str
print(type(lista))      # <class 'list'>

print(texto.upper())    # upper é um método dos objetos str → OLÁ MUNDO
lista.append(0)         # append é um método dos objetos list
lista.sort()
print(lista)            # [0, 1, 2, 3]
````

> 🤖 **Ligação à IA:** no pandas, `df.head()` é um método de um objeto `DataFrame`; em machine learning, `modelo.fit(dados)` é um método de um objeto "modelo". Perceber classes é perceber como estas bibliotecas funcionam.

<a id="sec-4-2"></a>

## 4.2. Criar uma classe: `class`, `__init__` e `self`

- `class Aluno:` define a classe (por convenção, nomes de classes começam por maiúscula).
- `__init__` é um método especial que corre **automaticamente** quando crias um objeto. Serve para preparar os atributos.
- `self` é **o próprio objeto** que está a ser criado/usado. `self.nome = nome` guarda o nome *dentro* desse objeto.

````python
class Aluno:
    def __init__(self, nome, idade):
        self.nome = nome      # atributo
        self.idade = idade    # atributo
        self.notas = []       # todos os alunos começam sem notas


ana = Aluno("Ana", 20)     # cria um objeto (o Python passa o self sozinho)
rui = Aluno("Rui", 22)

print(ana.nome, ana.idade)   # Ana 20
print(rui.nome, rui.idade)   # Rui 22
print(type(ana))             # <class '__main__.Aluno'>
````

Cada objeto tem os **seus próprios** atributos. Alterar um não mexe no outro:

````python
ana.idade = 21
print(ana.idade, rui.idade)   # 21 22
````

## 4.3. Métodos

Um **método** é uma função definida dentro da classe. Recebe sempre `self` como primeiro parâmetro, para poder usar os atributos do objeto.

````python
class Aluno:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade
        self.notas = []

    def adicionar_nota(self, nota):
        self.notas.append(nota)

    def media(self):
        if not self.notas:
            return 0
        return sum(self.notas) / len(self.notas)


ana = Aluno("Ana", 20)
ana.adicionar_nota(15)
ana.adicionar_nota(18)
ana.adicionar_nota(17)

print(ana.notas)            # [15, 18, 17]
print(round(ana.media(), 2))  # 16.67
````

> 📝 **Nota:** `ana.media()` é o mesmo que `Aluno.media(ana)` — é por isso que o `self` existe: é a forma de o método saber **de que aluno** se trata.

## 4.4. `__str__` e `__repr__`

Experimenta imprimir um objeto: a saída não é nada útil.

````python
print(ana)   # algo como <__main__.Aluno object at 0x...>
````

Os métodos especiais (com dois `_` de cada lado, chamados *dunder*) permitem personalizar isto:

- `__str__` → texto "bonito" para pessoas, usado pelo `print()` e `str()`.
- `__repr__` → texto "técnico" para programadores, usado quando o objeto aparece dentro de uma lista ou no fim de uma célula. Idealmente, parece-se com o código que cria o objeto.

````python
class Ponto:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"

    def __repr__(self):
        return f"Ponto(x={self.x}, y={self.y})"


p = Ponto(1, 2)
print(p)                      # usa __str__  → (1, 2)
print([p, Ponto(3, 4)])       # dentro de uma lista usa __repr__ → [Ponto(x=1, y=2), Ponto(x=3, y=4)]
````

## 4.5. Atributos de classe vs. atributos de instância

- **Atributo de instância** (`self.algo`): cada objeto tem o seu valor.
- **Atributo de classe** (definido diretamente dentro da `class`): é **partilhado** por todos os objetos.

````python
class Carro:
    rodas = 4   # atributo de classe: igual para todos os carros

    def __init__(self, marca, cor):
        self.marca = marca   # atributos de instância: cada carro tem os seus
        self.cor = cor


c1 = Carro("Renault", "azul")
c2 = Carro("Fiat", "vermelho")

print(c1.marca, c1.cor, c1.rodas)   # Renault azul 4
print(c2.marca, c2.cor, c2.rodas)   # Fiat vermelho 4
print(Carro.rodas)                  # também se acede pela classe → 4
````

## 4.6. Classes + exceções: a `ContaBancaria`

Vamos juntar tudo: uma classe cujos métodos **lançam `ValueError`** quando recebem valores inválidos. Assim, é impossível pôr a conta num estado absurdo (saldo negativo, depósitos de -50 €…).

````python
class ContaBancaria:
    banco = "Banco Python"   # atributo de classe

    def __init__(self, titular, saldo=0):
        self.titular = titular
        self.saldo = saldo

    def depositar(self, valor):
        if valor <= 0:
            raise ValueError("O valor a depositar tem de ser positivo.")
        self.saldo += valor

    def levantar(self, valor):
        if valor <= 0:
            raise ValueError("O valor a levantar tem de ser positivo.")
        if valor > self.saldo:
            raise ValueError(f"Saldo insuficiente: só tens {self.saldo} €.")
        self.saldo -= valor

    def __str__(self):
        return f"Conta de {self.titular} ({self.banco}): {self.saldo} €"


conta = ContaBancaria("Marta", 100)
conta.depositar(50)
conta.levantar(30)
print(conta)   # Conta de Marta (Banco Python): 120 €
````

````python
for operacao, valor in [("depositar", -20), ("levantar", 500), ("levantar", 20)]:
    try:
        if operacao == "depositar":
            conta.depositar(valor)
        else:
            conta.levantar(valor)
        print(f"✅ {operacao} {valor} € → saldo {conta.saldo} €")
    except ValueError as erro:
        print(f"❌ {operacao} {valor} €: {erro}")
````

## 4.7. Herança e `super()`

**Herança** permite criar uma classe nova a partir de outra: a classe "filha" recebe tudo o que a "mãe" tem e pode acrescentar ou alterar coisas.

Uma `ContaPoupanca` **é uma** `ContaBancaria`, mas com juros. `super().__init__(...)` chama o `__init__` da classe mãe, para não repetirmos código.

````python
class ContaPoupanca(ContaBancaria):   # herda de ContaBancaria
    def __init__(self, titular, saldo=0, taxa_juro=0.02):
        super().__init__(titular, saldo)   # a mãe trata do titular e do saldo
        self.taxa_juro = taxa_juro         # só a poupança tem taxa de juro

    def aplicar_juros(self):
        juros = round(self.saldo * self.taxa_juro, 2)
        self.depositar(juros)   # reutiliza o método herdado
        return juros


poupanca = ContaPoupanca("Tiago", 1000, taxa_juro=0.03)
print(poupanca.aplicar_juros())   # 30.0
print(poupanca)                   # __str__ herdado → Conta de Tiago (Banco Python): 1030.0 €

poupanca.levantar(30)             # levantar também foi herdado
print(poupanca.saldo)             # 1000.0
print(isinstance(poupanca, ContaBancaria))   # uma poupança É uma conta → True
````

## 4.8. Objetos dentro de objetos: `Turma`

Objetos podem guardar outros objetos. Uma `Turma` tem uma lista de `Aluno` (a classe da secção 4.3):

````python
class Turma:
    def __init__(self, nome):
        self.nome = nome
        self.alunos = []

    def adicionar(self, aluno):
        self.alunos.append(aluno)

    def melhor_aluno(self):
        melhor = None
        for aluno in self.alunos:
            if melhor is None or aluno.media() > melhor.media():
                melhor = aluno
        return melhor


turma = Turma("Python 101")

for nome, notas in [("Ana", [15, 18, 17]), ("Rui", [12, 14]), ("Inês", [19, 17, 18])]:
    aluno = Aluno(nome, 20)
    for nota in notas:
        aluno.adicionar_nota(nota)
    turma.adicionar(aluno)

for aluno in turma.alunos:
    print(f"{aluno.nome:<5} média {aluno.media():.1f}")

print("Melhor aluno:", turma.melhor_aluno().nome)   # Inês
````

## 4.9. O atalho moderno: `@dataclass`

Muitas classes servem sobretudo para **guardar dados**. Escrever o `__init__` e o `__repr__` à mão para elas é repetitivo. O decorador `@dataclass` gera-os por ti, a partir de uma lista de atributos com o seu tipo:

````python
from dataclasses import dataclass

@dataclass
class Produto:
    nome: str
    preco: float
    quantidade: int = 1   # valor por omissão

    def total(self):
        return self.preco * self.quantidade


cafe = Produto("Café", 1.5, 3)
print(cafe)                 # Produto(nome='Café', preco=1.5, quantidade=3)
print(cafe.total())         # 4.5
print(Produto("Chá", 2.0))  # Produto(nome='Chá', preco=2.0, quantidade=1)
print(cafe == Produto("Café", 1.5, 3))   # compara pelos dados → True
````

> 📝 **Nota:** `nome: str` é uma *anotação de tipo*: documenta o que se espera, mas o Python **não a verifica** — `Produto(123, "barato")` não dá erro. Ferramentas como o VS Code usam estas anotações para te avisar de enganos.

import random
from collections import Counter

def lancar_dados(n, seed=None):
    random.seed(seed)
    lancamentos = [random.randint(1, 6) for _ in range(n)]
    return Counter(lancamentos)

class Retangulo:
    def __init__(self, largura, altura):
        if largura <= 0 or altura <= 0:
            raise ValueError("As medidas têm de ser positivas.")
        self.largura = largura
        self.altura = altura

    def area(self):
        return self.largura * self.altura

    def perimetro(self):
        return 2 * (self.largura + self.altura)

    def e_quadrado(self):
        return self.largura == self.altura


r = Retangulo(3, 4)
print(r.area())                         # 12
print(r.perimetro())                    # 14
print(r.e_quadrado())                   # False
print(Retangulo(5, 5).e_quadrado())     # True

try:
    Retangulo(-1, 2)
except ValueError as erro:
    print("Erro:", erro)                # Erro: As medidas têm de ser positivas.

Validar no `__init__` garante que **nunca** existe um retângulo com medidas inválidas: o objeto nem chega a ser criado.

```python
class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self.preco = preco


class Carrinho:
    def __init__(self):
        self.itens = {}   # {nome: (produto, quantidade)}

    def adicionar(self, produto, quantidade=1):
        self.itens[produto.nome] = (produto, quantidade)

    def remover(self, nome):
        if nome not in self.itens:
            raise ValueError(f"O produto '{nome}' não está no carrinho.")
        del self.itens[nome]

    def total(self):
        soma = 0
        for produto, quantidade in self.itens.values():
            soma += produto.preco * quantidade
        return round(soma, 2)


carrinho = Carrinho()
carrinho.adicionar(Produto("Pão", 0.5), 4)
carrinho.adicionar(Produto("Leite", 1.2))
print(carrinho.total())   # 3.2

carrinho.remover("Pão")
print(carrinho.total())   # 1.2

try:
    carrinho.remover("Chocolate")
except ValueError as erro:
    print("Erro:", erro)  # Erro: O produto 'Chocolate' não está no carrinho.
```

O `Carrinho` guarda objetos `Produto` e usa os seus atributos (`produto.preco`). Arredondamos o total porque somas de decimais podem dar valores como `3.2000000000000006`. **Desafio extra:** se adicionares um produto que já está no carrinho, soma as quantidades em vez de substituir.
