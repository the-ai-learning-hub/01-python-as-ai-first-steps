# 2. O que são estruturas de dados?

Até agora cada variável guardava **um** valor (um número, um texto…). Mas os programas reais lidam com **coleções**: uma lista de alunos, as palavras de um texto, os dados de um cliente.

Python tem quatro estruturas de dados base:

| Tipo | Exemplo | Numa frase |
|---|---|---|
| `list` | `[1, 2, 3]` | Sequência ordenada que podes alterar |

| `dict` | `{"nome": "Ana"}` | Pares **chave → valor** |

# 3. Listas

Uma **lista** é uma sequência ordenada de valores. Escreve-se com parênteses retos `[]` e os elementos separados por vírgulas. Pode ter valores de tipos diferentes e valores repetidos.

<a id="sec-3-1"></a>

## 3.1. Criar e aceder a elementos

````python
# Lista de números
numeros = [1, 2, 3, 4, 5]

# Lista de strings
frutas = ["maçã", "banana", "laranja"]

# Lista mista (tipos diferentes)
mista = [1, "texto", 3.14, True]

# Lista vazia
vazia = []

print(frutas)  # Saída: ['maçã', 'banana', 'laranja']
````

Cada elemento tem um **índice** (posição), que começa em **0**:

````python
print(frutas[0])  # Saída: maçã
print(mista[1])   # Saída: texto
````

Os **índices negativos** contam a partir do fim, começando em `-1`:

````python
print(frutas[-1])  # Saída: laranja
print(frutas[-2])  # Saída: banana
````

> ⚠️ **Cuidado:** se pedires um índice que não existe, o Python dá um `IndexError`. A lista `frutas` tem 3 elementos, por isso os índices válidos são 0, 1 e 2 (esta célula falha de propósito).

````python
print(frutas[3])  # IndexError: list index out of range
````

<a id="sec-3-2"></a>

## 3.2. Alterar, adicionar e remover

As listas são **mutáveis**: podes alterar os elementos depois de as criares.

````python
frutas = ["maçã", "banana", "laranja"]

frutas[0] = "morango"       # substituir o elemento na posição 0
print(frutas)  # Saída: ['morango', 'banana', 'laranja']
````

**➕ Adicionar:** `append()` junta ao fim; `insert(posição, valor)` insere numa posição.

````python
frutas.append("kiwi")
print(frutas)  # Saída: ['morango', 'banana', 'laranja', 'kiwi']

frutas.insert(1, "ananás")
print(frutas)  # Saída: ['morango', 'ananás', 'banana', 'laranja', 'kiwi']
````

**➖ Remover:** `remove(valor)` remove pelo valor; `pop()` remove e **devolve** o último (ou o da posição indicada); `del` remove por índice.

````python
frutas.remove("banana")
print(frutas)  # Saída: ['morango', 'ananás', 'laranja', 'kiwi']

ultima = frutas.pop()
print(ultima)  # Saída: kiwi
print(frutas)  # Saída: ['morango', 'ananás', 'laranja']

del frutas[0]
print(frutas)  # Saída: ['ananás', 'laranja']
````

**🍉 Concatenar:** o operador `+` junta duas listas numa nova lista.

````python
lista1 = ["maçã", "banana"]
lista2 = ["laranja", "pera"]
todas_frutas = lista1 + lista2

print(todas_frutas)  # Saída: ['maçã', 'banana', 'laranja', 'pera']
````

## 3.3. `len`, `min`, `max`, `sum` e `in`

Há funções que funcionam com qualquer lista e que vais usar constantemente:

````python
notas = [14, 17, 9, 12, 18]

print(len(notas))  # Saída: 5
print(min(notas))  # Saída: 9
print(max(notas))  # Saída: 18
print(sum(notas))  # Saída: 70

media = sum(notas) / len(notas)
print(media)       # Saída: 14.0
````

O operador `in` verifica se um valor **existe** na lista (devolve `True` ou `False`):

````python
cientistas = ["Newton", "Einstein", "Curie", "Tesla"]

print("Curie" in cientistas)       # Saída: True
print("Darwin" in cientistas)      # Saída: False
print("Darwin" not in cientistas)  # Saída: True
````

## 3.4. Ordenar e inverter

Há duas formas de ordenar:

- `sorted(lista)` → **devolve uma lista nova** ordenada e deixa a original intacta.
- `lista.sort()` → ordena **a própria lista** e devolve `None`.

````python
numeros = [5, 2, 9, 1]

ordenados = sorted(numeros)
print(ordenados)  # Saída: [1, 2, 5, 9]
print(numeros)    # Saída: [5, 2, 9, 1]

numeros.sort()
print(numeros)    # Saída: [1, 2, 5, 9]

# Ordem decrescente
print(sorted(numeros, reverse=True))  # Saída: [9, 5, 2, 1]
````

> ⚠️ **Cuidado:** `.sort()` não devolve nada (devolve `None`). Um erro clássico é escrever `numeros = numeros.sort()` — e perder a lista!

````python
numeros = [5, 2, 9, 1]
resultado = numeros.sort()
print(resultado)  # Saída: None
````

**📍 Posição de um elemento** com `index()` e **🔄 inverter** com `reverse()` ou com o *slice* `[::-1]`:

````python
cores = ["vermelho", "verde", "azul"]
print(cores.index("azul"))  # Saída: 2

lista = [10, 20, 30, 40]
print(lista[::-1])  # Saída: [40, 30, 20, 10]
print(lista)        # Saída: [10, 20, 30, 40]

lista.reverse()     # inverte a própria lista
print(lista)        # Saída: [40, 30, 20, 10]
````

## 3.5. *Slicing*

O *slicing* extrai uma parte da lista com a notação `lista[início:fim:passo]` — exatamente como nas strings:

- `início` → índice onde começa (incluído; por omissão, 0);
- `fim` → índice onde termina (**excluído**; por omissão, o fim da lista);
- `passo` → de quantos em quantos (opcional; por omissão, 1).

````python
cientistas = ["Newton", "Einstein", "Curie", "Tesla", "Feynman", "Hawking"]

# Os 3 primeiros
print(cientistas[0:3])  # Saída: ['Newton', 'Einstein', 'Curie']

# Do 2.º ao 4.º elemento (índices 1, 2 e 3 — o 4 fica de fora)
print(cientistas[1:4])  # Saída: ['Einstein', 'Curie', 'Tesla']

# Os 2 últimos
print(cientistas[-2:])  # Saída: ['Feynman', 'Hawking']

# De 2 em 2 (salta um elemento)
print(cientistas[::2])  # Saída: ['Newton', 'Curie', 'Feynman']
````

## 3.6. Copiar listas (cuidado!)

Escrever `b = a` **não copia** a lista: cria apenas um segundo nome para a **mesma** lista. Alterar uma é alterar a outra!

````python
a = [1, 2, 3]
b = a          # b é outro nome para a mesma lista
b.append(4)

print(a)  # Saída: [1, 2, 3, 4]
print(b)  # Saída: [1, 2, 3, 4]
````

A lista `a` também mudou! Para teres uma **cópia independente**, usa `.copy()` (ou `a[:]`):

````python
a = [1, 2, 3]
c = a.copy()   # cópia nova
c.append(99)

print(a)  # Saída: [1, 2, 3]
print(c)  # Saída: [1, 2, 3, 99]
````

> ⚠️ **Cuidado:** isto também acontece quando passas uma lista a uma função: se a função a alterar com `append`, `remove`, etc., a lista original muda. Se não quiseres isso, trabalha numa cópia.

## 3.7. Do texto à lista: `split()`

Uma tarefa muito comum ao analisar texto — na filologia ou no processamento de linguagem natural — é separar uma frase em palavras individuais, chamadas *tokens*.

O método `.split()` divide uma **string** numa **lista**. Por omissão, separa nos espaços; também podes indicar outro separador, como `","`. O inverso é `" ".join(lista)`.

````python
texto_antigo = "In principio erat Verbum et Verbum erat apud Deum"
palavras = texto_antigo.split()

print(palavras)
print("Total de palavras:", len(palavras))     # Saída: Total de palavras: 9
print("Palavras únicas:", len(set(palavras)))  # Saída: Palavras únicas: 7
````

````python
texto_lista = "maçã,banana,laranja,maçã,pera,banana"
frutas = texto_lista.split(",")  # a vírgula é o separador

print(frutas)                              # Saída: ['maçã', 'banana', 'laranja', 'maçã', 'pera', 'banana']
print("Total de frutas:", len(frutas))     # Saída: Total de frutas: 6
print("Frutas únicas:", len(set(frutas)))  # Saída: Frutas únicas: 4

# E o caminho inverso: lista -> texto
print(" | ".join(frutas))  # Saída: maçã | banana | laranja | maçã | pera | banana
````

> 🤖 **Ligação à IA:** antes de um modelo de linguagem "ler" um texto, este é partido em *tokens* — a mesma ideia do `split()`, só que mais sofisticada.

> 📝 **Nota:** `set(palavras)` cria um conjunto sem duplicados — vais ver os sets na [secção 5](#sec-5).

# 4. Tuplos

Os **tuplos** são parecidos com as listas — ordenados, com índices, aceitam duplicados — mas são **imutáveis**: depois de criados, não podes adicionar, remover nem alterar elementos.

São ideais para dados que **não devem mudar**: coordenadas, datas, cores RGB…

<a id="sec-4-1"></a>

## 4.1. Criar e aceder

````python
lisboa = (38.72, -9.14)              # coordenadas (latitude, longitude)
cores = "vermelho", "verde", "azul"  # também é válido sem parênteses

print(lisboa)      # Saída: (38.72, -9.14)
print(cores)       # Saída: ('vermelho', 'verde', 'azul')
print(lisboa[0])   # Saída: 38.72
print(cores[-1])   # Saída: azul
print(len(cores))  # Saída: 3
````

Para criar um tuplo com **um só** elemento, precisas de uma vírgula no fim — sem ela, os parênteses são só parênteses:

````python
tuplo_unico = (5,)
print(type(tuplo_unico))  # Saída: <class 'tuple'>

nao_e_tuplo = (5)
print(type(nao_e_tuplo))  # Saída: <class 'int'>
````

<a id="sec-4-2"></a>

## 4.2. Tuplos são imutáveis

Se tentares alterar um elemento, o Python dá um `TypeError` (esta célula falha de propósito):

````python
lisboa[0] = 40.0  # TypeError: 'tuple' object does not support item assignment
````

> 💡 **Dica:** se precisares mesmo de alterar, converte numa lista, altera e volta a converter: `list(tuplo)` / `tuple(lista)`. Os tuplos só têm dois métodos: `.count(valor)` e `.index(valor)`.

<a id="sec-4-3"></a>

## 4.3. Desempacotar tuplos

**Desempacotar** é atribuir cada elemento de um tuplo (ou lista) a uma variável, numa só linha:

````python
lisboa = (38.72, -9.14)
latitude, longitude = lisboa

print(latitude)   # Saída: 38.72
print(longitude)  # Saída: -9.14
````

Um truque muito usado: **trocar** o valor de duas variáveis sem variável auxiliar:

````python
a = "café"
b = "chá"
a, b = b, a

print(a, b)  # Saída: chá café
````

Quando uma função faz `return x, y`, está na verdade a devolver um **tuplo** — que podes desempacotar logo:

````python
def minimo_e_maximo(valores):
    return min(valores), max(valores)

menor, maior = minimo_e_maximo([7, 3, 9, 1])
print(menor, maior)  # Saída: 1 9
````

# 5. Sets (conjuntos)

Um **set** guarda valores **únicos** e **sem ordem**: não aceita duplicados e não tem índices (`animais[0]` dá erro). É ótimo para eliminar repetidos e para comparar coleções.

<a id="sec-5-1"></a>

## 5.1. Criar sets e eliminar duplicados

````python
animais = {"gato", "cão", "pássaro", "gato"}

print(animais)           # os elementos podem aparecer por qualquer ordem
print(len(animais))      # Saída: 3
print("cão" in animais)  # Saída: True
````

O `"gato"` repetido desapareceu: o set só tem 3 elementos.

> ⚠️ **Cuidado:** um set **não tem ordem garantida**. Se executares a célula acima noutro computador (ou noutra sessão), os elementos podem aparecer por outra ordem. Quando precisares de uma ordem, usa `sorted(meu_set)`, que devolve uma lista ordenada.

Também podes criar um set com a função `set()`, a partir de qualquer sequência — por exemplo, as letras de uma palavra:

````python
letras = set("filologia")
print(sorted(letras))  # Saída: ['a', 'f', 'g', 'i', 'l', 'o']
print(len(letras))     # Saída: 6
````

A palavra tem 9 letras, mas só 6 diferentes. A utilização mais comum de um set: **eliminar duplicados de uma lista**, convertendo-a (e voltando a lista, se precisares):

````python
visitas = ["Ana", "Rui", "Ana", "Marta", "Rui", "Ana"]

unicos = set(visitas)
print(len(unicos))     # Saída: 3
print(sorted(unicos))  # Saída: ['Ana', 'Marta', 'Rui']
````

> ⚠️ **Cuidado:** `{}` cria um **dicionário** vazio, não um set! Para um set vazio usa `set()`.

````python
print(type({}))     # Saída: <class 'dict'>
print(type(set()))  # Saída: <class 'set'>
````

<a id="sec-5-2"></a>

## 5.2. Operações entre sets

Os sets permitem as operações dos conjuntos da matemática. Usamos `sorted()` só para o resultado aparecer sempre pela mesma ordem.

````python
frutas = {"maçã", "banana", "pera"}
doces = {"banana", "chocolate", "bolo"}

# União – elementos que estão em pelo menos um dos sets
print(sorted(frutas | doces))  # Saída: ['banana', 'bolo', 'chocolate', 'maçã', 'pera']

# Interseção – elementos em comum
print(sorted(frutas & doces))  # Saída: ['banana']

# Diferença – elementos de 'frutas' que não estão em 'doces'
print(sorted(frutas - doces))  # Saída: ['maçã', 'pera']
````

Também existem as versões em método: `a.union(b)`, `a.intersection(b)` e `a.difference(b)`.

<a id="sec-5-3"></a>

## 5.3. Métodos úteis

````python
numeros = {1, 2, 3, 2}
print(numeros)  # Saída: {1, 2, 3}

numeros.add(4)       # adiciona um elemento
print(numeros)  # Saída: {1, 2, 3, 4}

numeros.discard(2)   # remove (sem erro se não existir)
numeros.discard(10)  # não existe -> não acontece nada
print(numeros)  # Saída: {1, 3, 4}

numeros.clear()      # remove todos os elementos
print(numeros)  # Saída: set()
````

> 📝 **Nota:** também existe `.remove(valor)`, mas esse dá `KeyError` se o valor não existir. Na dúvida, usa `.discard()`.

# 6. Dicionários

Um **dicionário** guarda pares **chave → valor**. Em vez de acederes por posição (como nas listas), acedes pela **chave** — tal como procuras uma palavra num dicionário para obter o seu significado.

As chaves são **únicas** (normalmente strings) e os valores podem ser de qualquer tipo.

<a id="sec-6-1"></a>

## 6.1. Criar e aceder

````python
pessoa = {
    "nome": "Ana",
    "idade": 28,
    "cidade": "Lisboa",
}

print(pessoa)           # Saída: {'nome': 'Ana', 'idade': 28, 'cidade': 'Lisboa'}
print(pessoa["nome"])   # Saída: Ana
print(pessoa["idade"])  # Saída: 28
print(len(pessoa))      # Saída: 3
````

Pedir uma chave que não existe dá `KeyError` (esta célula falha de propósito):

````python
print(pessoa["email"])  # KeyError: 'email'
````

Para aceder com segurança, usa `.get(chave, valor_por_omissão)` — se a chave não existir, devolve o valor por omissão (ou `None`). E `in` verifica se uma **chave** existe:

````python
print(pessoa.get("cidade"))              # Saída: Lisboa
print(pessoa.get("email"))               # Saída: None
print(pessoa.get("email", "sem email"))  # Saída: sem email

print("nome" in pessoa)  # Saída: True
print("Ana" in pessoa)   # Saída: False
````

Repara no último: `"Ana" in pessoa` dá `False` porque o `in` procura nas **chaves**, não nos valores.

<a id="sec-6-2"></a>

## 6.2. Adicionar, modificar e remover

Para adicionar **ou** modificar, a sintaxe é a mesma: `dicionario[chave] = valor`. Se a chave já existe, o valor é substituído.

````python
pessoa["profissao"] = "Engenheira"  # chave nova -> adiciona
pessoa["idade"] = 29                # chave existente -> atualiza

print(pessoa)  # Saída: {'nome': 'Ana', 'idade': 29, 'cidade': 'Lisboa', 'profissao': 'Engenheira'}
````

Para remover, usa `pop(chave)` (que devolve o valor removido) ou `del`:

````python
cidade = pessoa.pop("cidade")
print(cidade)  # Saída: Lisboa

del pessoa["profissao"]
print(pessoa)  # Saída: {'nome': 'Ana', 'idade': 29}
````

<a id="sec-6-3"></a>

## 6.3. `keys()`, `values()` e `items()`

- `.keys()` → todas as chaves
- `.values()` → todos os valores
- `.items()` → todos os pares `(chave, valor)`, como tuplos

Atenção: não devolvem listas, mas sim **vistas** (`dict_keys`, `dict_values`, `dict_items`) — objetos que "espreitam" o dicionário. Se precisares de uma lista verdadeira, usa `list(...)`.

````python
livro = {"titulo": "Os Lusíadas", "autor": "Camões", "ano": 1572}

print(livro.keys())    # Saída: dict_keys(['titulo', 'autor', 'ano'])
print(livro.values())  # Saída: dict_values(['Os Lusíadas', 'Camões', 1572])
print(livro.items())   # Saída: dict_items([('titulo', 'Os Lusíadas'), ('autor', 'Camões'), ('ano', 1572)])

chaves = list(livro.keys())
print(chaves)          # Saída: ['titulo', 'autor', 'ano']
````

> 📝 **Nota:** as vistas acompanham o dicionário: se adicionares uma chave depois, ela aparece na vista sem teres de chamar `.keys()` outra vez. Desde o Python 3.7, os dicionários mantêm a **ordem de inserção**.

<a id="sec-6-4"></a>

## 6.4. Contar palavras com `Counter`

Contar quantas vezes aparece cada elemento é tão comum que o Python traz uma ferramenta pronta: `Counter`, do módulo `collections`. Um `Counter` é um dicionário em que os valores são contagens.

````python
from collections import Counter

frase = "A vida é bela e a vida é curta"
palavras = frase.lower().split()  # lower() para "A" e "a" contarem como a mesma palavra

contagem = Counter(palavras)
print(contagem)  # Saída: Counter({'a': 2, 'vida': 2, 'é': 2, 'bela': 1, 'e': 1, 'curta': 1})

print(contagem["vida"])         # Saída: 2
print(contagem.most_common(1))  # Saída: [('a', 2)]
````

resposta = {
    "total": 2,
    "utilizadores": [
        {"nome": "Ana", "cidade": "Lisboa", "interesses": ["IA", "música"]},
        {"nome": "Rui", "cidade": "Porto", "interesses": ["futebol", "dados"]},
    ],
}


def apresentar(resposta, posicao):
    utilizador = resposta["utilizadores"][posicao]
    nome = utilizador["nome"]
    cidade = utilizador["cidade"]
    interesse = utilizador["interesses"][0]
    return f"{nome} vive em {cidade} e gosta de {interesse}"


print(apresentar(resposta, 0))  # esperado: Ana vive em Lisboa e gosta de IA
print(apresentar(resposta, 1))  # esperado: Rui vive em Porto e gosta de futebol

# 📌 Resumo

| Conceito | Sintaxe | Exemplo |
|---|---|---|
| Criar lista | `[a, b, c]` | `frutas = ["maçã", "pera"]` |
| Aceder por índice | `lista[i]`, `lista[-1]` | `frutas[0]` → `'maçã'` |
| Adicionar / remover | `append`, `insert`, `remove`, `pop`, `del` | `frutas.append("kiwi")` |
| Tamanho, mínimo, máximo, soma | `len`, `min`, `max`, `sum` | `sum([1, 2, 3])` → `6` |
| Existe? | `x in coleção` | `"pera" in frutas` → `True` |
| Ordenar | `sorted(l)` (nova) / `l.sort()` (devolve `None`) | `sorted([3, 1, 2])` → `[1, 2, 3]` |
| *Slicing* | `lista[início:fim:passo]` | `[10, 20, 30][:2]` → `[10, 20]` |
| Copiar | `l.copy()` ou `l[:]` (não `b = a`) | `b = a.copy()` |
| Texto → lista → texto | `s.split(sep)` / `sep.join(l)` | `"a b".split()` → `['a', 'b']` |
| Tuplo | `(a, b)`, `(a,)` | `lisboa = (38.72, -9.14)` |
| Desempacotar / trocar | `x, y = tuplo` / `a, b = b, a` | `lat, lon = lisboa` |
| Set | `{a, b}`, `set(seq)`, `set()` vazio | `set([1, 1, 2])` → `{1, 2}` |
| Operações de sets | `\|`, `&`, `-` | `{1, 2} & {2, 3}` → `{2}` |
| Dicionário | `{chave: valor}` | `p = {"nome": "Ana"}` |

| Contar | `Counter(seq)` | `Counter("aab")` → `Counter({'a': 2, 'b': 1})` |
| Aninhar | `d["lista"][0]["chave"]` | `resposta["resultados"][0]["rotulo"]` |
